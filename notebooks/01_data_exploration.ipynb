```python
import pandas as pd

movies = pd.read_csv("../data/tmdb_5000_movies.csv")
credits = pd.read_csv("../data/tmdb_5000_credits.csv")

print("Movies shape:", movies.shape)
print("Credits shape:", credits.shape)
```

```python
print("MOVIES COLUMNS:")
print(movies.columns.tolist())

print("\nCREDITS COLUMNS:")
print(credits.columns.tolist())
```

```python
movies.head()
```

```python
print("MOVIES MISSING VALUES:")
print(movies.isnull().sum())

print("\nCREDITS MISSING VALUES:")
print(credits.isnull().sum())
```

```python
print("Duplicate rows in movies:", movies.duplicated().sum())
print("Duplicate rows in credits:", credits.duplicated().sum())
```

```python
print("MOVIES DATA TYPES:")
print(movies.dtypes)

print("\nCREDITS DATA TYPES:")
print(credits.dtypes)
```
