```python
columns_to_drop = [
    "homepage",
    "production_companies",
    "production_countries",
    "spoken_languages",
    "tagline",
    "budget",
    "revenue"
]

clean_movies = clean_movies.drop(columns=columns_to_drop)

print("Shape after removing unnecessary columns:", clean_movies.shape)
print("\nRemaining columns:")
print(clean_movies.columns.tolist())
```

```python
clean_movies.isnull().sum()
```

```python
clean_movies["overview"] = clean_movies["overview"].fillna("")

print("Missing overviews:", clean_movies["overview"].isnull().sum())
```

```python
clean_movies["release_date"] = pd.to_datetime(
    clean_movies["release_date"],
    errors="coerce"
)

print("Missing release dates:", clean_movies["release_date"].isnull().sum())
print("Data type:", clean_movies["release_date"].dtype)
```

```python
median_runtime = clean_movies["runtime"].median()

clean_movies["runtime"] = clean_movies["runtime"].fillna(median_runtime)

print("Median runtime:", median_runtime)
print("Missing runtime values:", clean_movies["runtime"].isnull().sum())
```

```python
import ast

print(type(clean_movies["genres"].iloc[0]))

genres_list = ast.literal_eval(clean_movies["genres"].iloc[0])

print(type(genres_list))
print(genres_list)
```

```python
def extract_names(text):
    items = ast.literal_eval(text)
    return [item["name"] for item in items]

print(extract_names(clean_movies["genres"].iloc[0]))
```

```python
clean_movies["genres"] = clean_movies["genres"].apply(extract_names)

clean_movies[["title", "genres"]].head()
```

```python
clean_movies["keywords"] = clean_movies["keywords"].apply(extract_names)

clean_movies[["title", "keywords"]].head()
```

```python
def extract_top_cast(text):
    cast = ast.literal_eval(text)
    return [actor["name"] for actor in cast[:5]]

print(extract_top_cast(credits["cast"].iloc[0]))
```

```python
credits["cast"] = credits["cast"].apply(extract_top_cast)

credits[["title", "cast"]].head()
```

```python
crew = ast.literal_eval(credits["crew"].iloc[0])

directors = [
    person["name"]
    for person in crew
    if person["job"] == "Director"
]

print(directors)
```

```python
def extract_director(text):
    crew = ast.literal_eval(text)
    
    for person in crew:
        if person["job"] == "Director":
            return person["name"]
    
    return ""
credits["director"] = credits["crew"].apply(extract_director)

credits[["title", "director"]].head()
```

```python
credits_clean = credits[
    ["movie_id", "cast", "director"]
].copy()

credits_clean.head()
```

```python
final_movies = clean_movies.merge(
    credits_clean,
    left_on="id",
    right_on="movie_id",
    how="inner"
)

print("Shape after merge:", final_movies.shape)
print(final_movies.columns.tolist())
```

```python
final_movies = final_movies.drop(columns=["movie_id"])

print("Final shape:", final_movies.shape)
print(final_movies.columns.tolist())
```

```python
final_movies[["title", "genres", "keywords", "cast", "director"]].head()
```

```python
def list_to_string(items):
    return " ".join(items)

final_movies["genres"] = final_movies["genres"].apply(list_to_string)
final_movies["keywords"] = final_movies["keywords"].apply(list_to_string)
final_movies["cast"] = final_movies["cast"].apply(list_to_string)

final_movies[["title", "genres", "keywords", "cast"]].head()
```

```python
final_movies["tags"] = (
    final_movies["overview"].fillna("") + " " +
    final_movies["genres"].fillna("") + " " +
    final_movies["keywords"].fillna("") + " " +
    final_movies["cast"].fillna("") + " " +
    final_movies["director"].fillna("")
)

final_movies[["title", "tags"]].head()
```

```python
print("Missing tags:", final_movies["tags"].isnull().sum())
print("Tags type:", type(final_movies["tags"].iloc[0]))
print("Number of movies:", len(final_movies))

print("\nAvatar tags:")
print(final_movies.loc[final_movies["title"] == "Avatar", "tags"].iloc[0])
```

```python
print("Missing tags:", movies["tags"].isnull().sum())
print("Tags type:", type(movies["tags"].iloc[0]))

print("\nFirst movie tags:")
print(movies["tags"].iloc[0])
```

```python
movies["tags"] = movies["tags"].str.lower()

print(movies["tags"].iloc[0])
```

```python
movies["tags"] = movies["tags"].apply(
    lambda x: x.replace(" ", "")
)

print(movies["tags"].iloc[0])
```

```python
from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer(
    max_features=5000,
    stop_words="english"
)

vectors = cv.fit_transform(movies["tags"])

print("Vector shape:", vectors.shape)
```

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity = cosine_similarity(vectors)

print("Similarity matrix shape:", similarity.shape)
```

```python
def recommend(movie_title):
    movie_index = movies[movies["title"] == movie_title].index[0]

    similarity_scores = list(enumerate(similarity[movie_index]))

    similarity_scores = sorted(
        similarity_scores,
        key=lambda x: x[1],
        reverse=True
    )

    top_movies = similarity_scores[1:6]

    for index, score in top_movies:
        print(movies.iloc[index]["title"])
```

```python
recommend("Avatar")
```

```python
print("Cast missing:", movies["cast"].isnull().sum())
print("Director missing:", movies["director"].isnull().sum())

print("\nCast type:", type(movies["cast"].iloc[0]))
print("Director type:", type(movies["director"].iloc[0]))
```

```python
movies["cast"] = movies["cast"].fillna("").str.replace(" ", "", regex=False)
movies["director"] = movies["director"].fillna("").str.replace(" ", "", regex=False)

print("Cast:", movies["cast"].iloc[0])
print("Director:", movies["director"].iloc[0])
```

```python
movies["tags"] = (
    movies["overview"].fillna("") + " " +
    movies["genres"].fillna("") + " " +
    movies["keywords"].fillna("") + " " +
    movies["cast"].fillna("").str.replace(
        r"(?<=[a-z])(?=[A-Z])", " ", regex=True
    ) + " " +
    movies["director"].fillna("")
)

movies["tags"] = movies["tags"].str.lower()

print(movies["tags"].iloc[0])
```

```python
cv = CountVectorizer(
    max_features=5000,
    stop_words="english"
)

vectors = cv.fit_transform(movies["tags"])

print("New vector shape:", vectors.shape)
```

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity = cosine_similarity(vectors)

print("Similarity matrix shape:", similarity.shape)
```

```python
def recommend(movie_title):
    movie_matches = movies[movies["title"].str.lower() == movie_title.lower()]

    if movie_matches.empty:
        print("Movie not found. Please check the title.")
        return

    movie_index = movie_matches.index[0]

    similarity_scores = list(enumerate(similarity[movie_index]))

    similarity_scores = sorted(
        similarity_scores,
        key=lambda x: x[1],
        reverse=True
    )

    top_movies = similarity_scores[1:6]

    for index, score in top_movies:
        print(
            movies.iloc[index]["title"],
            "→ Similarity:",
            round(score, 3)
        )
```
